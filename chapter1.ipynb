# Creating Data Frames

```python
freinds_data = {
    'Name': ["Rajeev","Rahul","Rakesh","Mukesh"],
    "Age": [25,15,45,85],
    "Insured":[True,False,False,True]
}
```

```python
freind_df = pd.DataFrame(freinds_data,index=['A','B','C','D'])
freind_df
```

# Renaming Columns In data Frame 

```python
freind_df.rename(columns={'Name':'Friends_Name', 'Age':'Freinds_Age','Insured':'Insurance'})
```
