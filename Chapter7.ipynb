# 🔷 1. Handling Missing Data (NaN)

```python
df = pd.read_csv("C:\Data Analytics\EXCEL ADVANCE\PROJECTS\Project -2 Hospital Room Analysis\Hospital Emergency Room Data.csv")
```

# 🟢 Detecting missing values

```python
df.isnull() # # Shows True where data is missing
```

```python
df.notnull()        # Shows True where data is present
```

# 🟢 Counting missing values

```python
df.isnull().sum() # Count NaN in each column
```

# 🟢 Removing missing values

```python
df.dropna(axis = 1,how='all')
```

```python
df.dropna(axis=1) # Drops columns with NaN
```

# 🟢 Filling missing values

```python
df.fillna(0) # Replace NaN with 0
```

```python
df.fillna("Unknown") # Replace NaN with a string
```

```python
df.columns
```

```python
df["Patient Satisfaction Score"].fillna(df['Patient Satisfaction Score'].mean())
```

# 🔷 2. Fixing Wrong Data Types

```python
df.dtypes
```

```python
df['Patient Admission Date'] = pd.to_datetime(df['Patient Admission Date'], format='mixed')
```

# 🔷 3. Removing Duplicates

# 🧹 Detect and Remove

```python
df.duplicated()
```

```python
df.drop_duplicates()
```

# 🔷 4. Replacing Values

```python
df['Patient First Inital'].replace("H","Rajeev Kumar", inplace=True)
```

```python
df.head()
```

```python
df['Patient First Inital'].replace({
    "Rajeev Kumar":"Rajeev",
    "X":"Ravi"
}, inplace= True)
```

```python
df.head()
```

# 🔷 5. String Operations and Cleaning

```python
df["Patient First Inital"] = df["Patient First Inital"].str.strip()
df["Patient Last Name"] = df["Patient Last Name"].str.lower()
df["Department Referral"] = df["Department Referral"].str.upper()
df["Patient Race"] = df["Patient Race"].str.replace("/", "|", regex=False)

df.head()
```

# 🔷 6. Renaming Columns

```python
df.rename(columns={'Patient First Inital':"First Name", 'Patient Last Name':"Last Name"},inplace=True)
df.head()
```

# 🧪 Real-Life Task for Practice:


```python
data = {
    "Name": ["Rajeev", "Rahul", None, "Rakesh", "Rakesh"],
    "Age": [25, None, 30, 45, 45],
    "Gender": ["M", "F", "M", "M", "M"]
}

df2 = pd.DataFrame(data)
```

```python
df2.isnull()
```

```python
df2['Name'].fillna("Unknown" , inplace=True)
df2['Age'].fillna(df2['Age'].mean(),inplace=True)
```

```python
df2
```

```python
df2.drop_duplicates()
```

```python
df2.rename(columns={"Gender":"Sex"})
```

```python
# Step 1: Fill missing values in 'Age' column
df2['Age'].fillna(df2['Age'].mean(), inplace=True)

# Step 2: Convert to int safely
df2['Age'] = df2['Age'].astype(int)
```
