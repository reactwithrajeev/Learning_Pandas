# 🚀 Task 2: Time Shifting Challenge
👉 Using the same dataframe (df you created just now):

🔵 Create a new column called 'Yesterday Sales'
🔵 In this column, shift the 'Sales' data by 1 day backward (means ➡️ today's "Yesterday Sales" should be yesterday's actual Sales)



```python
df['Yesterday Sales'] = df['Sales'].shift(1)
df
```

```python
import pandas as pd

data = {
    'Department': ['HR', 'HR', 'IT', 'IT', 'Finance', 'Finance'],
    'Gender': ['Male', 'Female', 'Male', 'Female', 'Male', 'Female'],
    'Salary': [50000, 52000, 60000, 61000, 70000, 72000],
    'Employee': ['E1', 'E2', 'E3', 'E4', 'E5', 'E6']
}
df = pd.DataFrame(data)
```

```python
df.groupby('Department')['Salary'].mean()
```

```python
df.pivot_table(index='Department',columns='Gender',values='Salary', aggfunc='mean', fill_value=0)
```

```python
pd.crosstab(index=df['Department'], columns=df['Gender'], values=df['Employee'], aggfunc='count')
```
