# 🚀 Task 5: Handling Missing Values
Create a new DataFrame with some missing values (use NaN).

Use the .isnull() function to find which values are missing.

Use .fillna() to replace missing values with "Unknown" for one column and the mean value of the column for another.

Drop any rows that have missing values using .dropna().

```python
import numpy as np
```

```python
friends2 = {
    "Name": ['Rajeev', np.nan, 'Shahank', np.nan],
    'Age': [25, np.nan, 30, np.nan],
    'City': ['Gurgaon', np.nan, 'Mumbai', np.nan]
}

df2 = pd.DataFrame(friends2)
df2
```

```python
df2.isnull
```

```python
df2['Name'].fillna("Unknown" , inplace=True)
df2['Age'].fillna(df2['Age'].mean(),inplace=True)
```

```python
df2
```

```python
df2.dropna(axis=1)
```

```python
df2.dropna()
```

# ✅ Task 11: Grouping and Aggregation
Use the following DataFrame:

```python
data = {
    "Department": ["Sales", "HR", "Sales", "HR", "Finance", "Sales", "Finance"],
    "Name": ["Raj", "Ravi", "Kiran", "Meena", "Ali", "Sneha", "Tom"],
    "Salary": [50000, 45000, 55000, 47000, 60000, 52000, 62000]
}

df = pd.DataFrame(data)
```

```python
df.groupby('Department')['Salary'].mean()
```

```python
df.groupby('Department')['Salary'].max()
```

```python
avg_salary = df.groupby('Department')['Salary'].mean().sort_values(ascending=False)
avg_salary
```
