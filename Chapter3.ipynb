# Selecting Rows by Index Label — 

# ➡️ What is loc[]?
# Label-based selection
# You can select one row, multiple rows, or rows + columns by their index label (like 'A', 'B', 'C' here).

```python
ins_data.index
```

```python
ins_data.loc[[400,450,500,550],['age','smoker']]
```

# ✅ 4️⃣ Selecting Rows by Index Position — iloc[]
# ➡️ What is iloc[]?
# Position-based selection
# Uses integer positions (like 0, 1, 2, 3) to pick rows and columns.

```python
ins_data.iloc[[100,101],[1,2]]
```

# ✅ 5️⃣ Filtering Rows Based on a Condition
# ➡️ Why Filter?
# Sometimes you need only the rows that match a rule (like "age greater than 40")

```python
ins_data[ins_data['sex']=='male']
```

```python
ins_data[(ins_data['age']>35) & (ins_data['smoker']=='yes')]
```

# ✅ 6️⃣ Selecting Specific Cells (Rows + Columns Together)
# You can select a single value by combining:
# loc[row_label, column_label]
# iloc[row_position, column_position]

```python
ins_data.loc[400,'bmi']
```

```python
ins_data.iloc[400,2]
```
