# 10. Pivot Table & Crosstab

```python
import pandas as pd
```

# Pivot Table Basics
A pivot table is a powerful tool for summarizing data. It allows you to transform data and aggregate it, making it easier to analyze trends and patterns. In Pandas, you can create a pivot table using the pivot_table() function.

# Syntax

df.pivot_table(values, index, columns, aggfunc)

# Where:

values: The column you want to aggregate.

index: The column(s) to group by (rows in the pivot table).

columns: The column(s) to use as columns in the pivot table.

aggfunc: The aggregation function to apply (e.g., mean, sum, count, etc.).


```python

data = {
    'Department': ['HR', 'Finance', 'HR', 'Finance', 'HR', 'Finance'],
    'Gender': ['M', 'F', 'M', 'F', 'M', 'M'],
    'Salary': [50000, None, 55000, 65000, 52000, 67000],
    'Experience': [5, 7, 6, 8, 4, 10]
}

df = pd.DataFrame(data)
df
```

```python
pivot = df.pivot_table(values='Salary',index='Department',aggfunc='mean')
pivot
```

```python
pivot = df.pivot_table(values=['Salary','Experience'],index='Department')
pivot
```

# Handling Missing Data in Pivot Tables
By default, Pandas excludes NaN values when creating a pivot table. However, you can control how missing values are handled using the fill_value argument. For example:


```python
pivot = df.pivot_table(values='Salary', index='Department', aggfunc='mean', fill_value=0)
print(pivot)
```

# Crosstab: Similar to Pivot Table, but for Categorical Data
A crosstab is a cross-tabulation of data, similar to a pivot table but used for categorical data. It helps you to quickly analyze the relationship between two categorical variables.

# Syntax:
pd.crosstab(index, columns, values=None, aggfunc=None)


```python
crosstab= pd.crosstab(df['Department'],df['Gender'])
crosstab
```

```python
crosstab = pd.crosstab(df['Department'], df['Gender'], values=df['Salary'], aggfunc='mean')
crosstab
```
