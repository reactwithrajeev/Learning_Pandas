# Chapter 8 -  GroupBy and Aggregations

```python
import pandas as pd
```

# 🔸 What is groupby()?

# 💻 Syntax:

# But we usually combine with aggregation:

-- df.groupby('column_name').agg_function()


```python
data = {
    "Store": ["A", "A", "B", "B", "C", "C"],
    "Product": ["Soap", "Shampoo", "Soap", "Shampoo", "Soap", "Shampoo"],
    "Sales": [100, 200, 150, 250, 130, 230]
}

df = pd.DataFrame(data)
```

# 🎯 Task: Group by Store and sum Sales

```python
df.groupby("Store")['Sales'].sum()
```

```python
df.groupby("Store")["Sales"].mean()     # Average sales per store
df.groupby("Store")["Sales"].max()      # Max sale per store
df.groupby("Store")["Sales"].min()      # Min sale per store
```

# 🧠 Multiple aggregations:

```python
df.groupby("Store")["Sales"].agg(["sum", "mean", "max", "min"])
```

# 🔄 Group by Multiple Columns:

```python
df.groupby(['Store','Product'])['Sales'].sum()
```
