# 🧠 What is groupby() in Pandas?

The groupby() method is used to split the data into groups based on some criteria, apply a function (like mean, sum, etc.), and combine the results.

This is also called:
Split → Apply → Combine

# 💡 Real-life Example:
Imagine a company has 100 employees across departments (like HR, Finance, IT), and you want to find the average salary in each department. groupby() makes that super easy.

# 💡 Advanced GroupBy - GroupBy with Multiple Columns

```python

data = {
    'Department': ['HR', 'IT', 'Finance', 'HR', 'IT', 'Finance', 'HR'],
    'Salary': [50000, 70000, 60000, 52000, 72000, 61000, 51000],
    'Experience': [3, 5, 6, 4, 7, 9, 2]
}

df = pd.DataFrame(data)
df
```

```python
df.groupby('Department').agg({
    'Salary':['mean','sum','max'],
    'Experience':['min','max']
})
```

```python
data = {
    'Department': ['HR', 'IT', 'Finance', 'HR', 'IT', 'Finance', 'HR', 'IT'],
    'Gender': ['M', 'M', 'F', 'F', 'M', 'F', 'M', 'F'],
    'Salary': [50000, 70000, 60000, 52000, 72000, 61000, 51000, 74000],
    'Experience': [3, 5, 6, 4, 7, 9, 2, 8]
}

df = pd.DataFrame(data)
df
```

```python
df.groupby(['Department','Gender']).agg({
    'Salary':'mean',
    'Experience':'sum'
})
```

```python
df.groupby('Gender').agg({
    'Experience': 'sum'
})
```

```python
df.groupby(['Department','Gender']).agg({
    'Salary':'mean',
    'Experience':'sum'
})
```
