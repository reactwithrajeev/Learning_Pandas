```python
import pandas as pd

data = {
    'Employee': ['John', 'Alice', 'Bob', 'Daisy', 'Ethan', 'Fiona'],
    'Department': ['HR', 'Finance', 'IT', 'Finance', 'HR', 'IT'],
    'Salary': [50000, 60000, 55000, 62000, 52000, 58000],
    'Gender': ['Male', 'Female', 'Male', 'Female', 'Male', 'Female']
}

df = pd.DataFrame(data)
```

```python
df
```

# 🎯 Your Task:
Use groupby() to find:

Average salary for each Department

```python
df.groupby('Department')['Salary'].mean()
```

# 🎯 Your Task:
Use groupby() to find:

Total salary and count of employees for each Department and Gender combination.

(Hint: Use agg() to apply multiple functions.)

```python
df.groupby(['Department','Gender']).agg({
    'Salary':'sum',
    'Employee':'count'
})
```

# 🎯 Your Task:
Create a pivot table that shows the average salary for each Department and Gender.

(Hint: use pivot_table() with index, columns, and values.)

```python
pivot = df.pivot_table(index='Department', columns='Gender',values='Salary', aggfunc='mean', fill_value=0)
pivot
```

# 📊 Pivot Table — Task 2
🎯 Your Task:
Create a pivot table that shows the total salary for each combination of Department and Gender.

This time, instead of average, calculate the sum.
(Use the same DataFrame.)

```python
pivot = df.pivot_table(index='Department', columns='Gender',values='Salary', aggfunc='sum', fill_value=0)
pivot
```

# 📊 Crosstab — Task 1
🎯 Your Task:
Create a crosstab that shows the count of employees by Department and Gender.

Use pd.crosstab() and just count how many employees are there in each combo.
Simple and clean.

```python
df
```

```python
crosstab = pd.crosstab(index=df['Department'],columns=df['Gender'])
crosstab
```

# 📌 Crosstab Task 2:
Using the same df, create a crosstab that shows the total Salary of employees by Department and Gender.

Hint: You’ll need to use values= and aggfunc= here 😉

```python
crosstab = pd.crosstab(index=df['Department'],columns=df['Gender'],values=df['Salary'],aggfunc='sum')
crosstab
```
