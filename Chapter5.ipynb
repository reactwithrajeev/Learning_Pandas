# Working with Missing Data (NaN) in Pandas

```python
import pandas as pd
import numpy as np
```

```python
data = {
    'Name': ['Rajeev', 'Rahul', 'Rakesh', 'Mukesh'],
    'Age': [25, np.nan, 35, 40],
    'City': ['Delhi', 'Mumbai', np.nan, 'Chennai']
}
```

```python
df = pd.DataFrame(data)
df
```

# 1️⃣ Check where NaN exists


```python
df.isnull()
```

# 2️⃣ Count total missing values

```python
df.isnull().sum()
```

# ✅ 1. Remove Missing Data

```python
df.dropna(inplace=True)
df
```

```python
df
```

# ✅ 2. Replace Missing Data

```python
df.fillna(0)
```

# 👉 Replace NaN with mean, median, or mode

```python
df['Age'].fillna(df['Age'].mean())
```

# 👉 Forward fill (copy previous value)

```python
df.fillna(method='ffill', inplace=True)
df
```

```python
df
```

```python
df.ffill()
```

```python
df.bfill()
```

# ✅ 3. Check If Any Missing Data Exists

```python
df.isnull().values.any()
```
